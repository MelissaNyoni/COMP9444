```python
%pip install huggingface_hub librosa numpy pandas joblib soundfile

from huggingface_hub import snapshot_download

REPO_ID = "marsyas/gtzan"

path = snapshot_download(
    repo_id=REPO_ID,
    repo_type="dataset",
    local_dir="./data",
)

print(f"Dataset downloaded to: {path}")
```

```python
import tarfile

with tarfile.open("data/data/genres.tar.gz", "r:gz") as tar:
    tar.extractall("data/extracted")

print("Extraction complete")
```

```python
import os

for root, dirs, files in os.walk("data/extracted"):
    print(root, "->", files[:3])  # show first 3 files per folder
    if len(files) > 3:
        print(f"   ... and {len(files) - 3} more")
```
